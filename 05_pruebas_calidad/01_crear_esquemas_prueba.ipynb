# Pruebas de calidad: crear esquemas de prueba

Crea el esquema `pruebas_calidad` en los cuatro catálogos y copia **solo la estructura** de las tablas de control, para que las corridas de prueba no ensucien `hd_etl_log`, `hd_deduplicacion_log` ni `hd_cuarentena_resumen` reales.

Es seguro correrlo más de una vez: no borra nada.

```python
CATALOGOS = ["brz_dataco", "slv_dataco", "gld_dataco", "metadata_dataco"]
ESQUEMA_PRUEBA = "pruebas_calidad"

for catalogo in CATALOGOS:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalogo}.{ESQUEMA_PRUEBA}")
    spark.sql(f"COMMENT ON SCHEMA {catalogo}.{ESQUEMA_PRUEBA} IS 'Datos sinteticos para probar reglas de calidad. No contiene datos de negocio.'")
    print(f"✓ Esquema {catalogo}.{ESQUEMA_PRUEBA}")
```

```python
# Solo estructura (sin filas) de las tablas de control que escriben los notebooks Silver.
TABLAS_CONTROL = ["hd_etl_log", "hd_deduplicacion_log", "hd_cuarentena_resumen"]

for tabla in TABLAS_CONTROL:
    spark.sql(f"""
        CREATE TABLE IF NOT EXISTS metadata_dataco.{ESQUEMA_PRUEBA}.{tabla}
        LIKE metadata_dataco.supply_chain.{tabla}
    """)
    print(f"✓ metadata_dataco.{ESQUEMA_PRUEBA}.{tabla}")

display(spark.sql(f"SHOW TABLES IN metadata_dataco.{ESQUEMA_PRUEBA}"))
```
